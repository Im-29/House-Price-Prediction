## **DATA CLEANING** ##

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_excel("C:/Users/LENOVO/Documents/3rd YEAR/SEM 2/FYP 2/draft code/SELANGOR.xlsx")

# Display first 5 rows
display(df.head())

# Check dataset information
display(df.info())

# Check missing values
display(df.isnull().sum())
```

**Standardize Attributes Name**

```python
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
    .str.replace(",", "")
)

display(df.columns)
```

**Fill Null Values**

```python
df = df.ffill()
display(df.isnull().sum())
```

**Handle Duplicate**

```python
display(df.shape)
df = df.drop_duplicates()
display(df.shape)
```

```python
df["transaction_price"] = (
    df["transaction_price"]
    .astype(str)
    .str.replace("RM", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace(".00", "", regex=False)
    .str.strip()
)

df["transaction_price"] = pd.to_numeric(
    df["transaction_price"],
    errors="coerce"
)

display(df["transaction_price"].head())
```

```python
area_columns = ["land_parcel_area", "main_floor_area"]

for col in area_columns:
    
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.extract(r'(\d+\.?\d*)')[0]
    )

    df[col] = pd.to_numeric(df[col], errors="coerce")

display(df[area_columns].head())
```

```python
df = df.dropna(subset=["transaction_price"])
```

**Drop Unimportant Attributes**

```python
df = df.drop(columns=["unit"])
df = df.drop(columns=["mukim"])
df = df.drop(columns=["scheme_name_area"])
df = df.drop(columns=["road_name"])
df = df.drop(columns=["unit_level"])
df = df.drop(columns=["land_parcel_area"])
display(df.head())
```

```python
display(df.info())
display(df.isnull().sum())
display(df.head())
```

```python
df.to_csv("cleaned_selangor_dataset.csv", index=False)

print("Dataset cleaned successfully!")
```

## **DATA PREPROCESSING** ##

```python
features = [
    "property_type",
    "district",
    "month_year_of_transaction_date",
    "tenure",
    "main_floor_area"
]

X = df[features]

y = df["transaction_price"]
```

```python
display(X.head(30))
display(y.head(30))
```

```python
print(X.isnull().sum())
```

```python
X = pd.get_dummies(X)
display(X.head(10))
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

X = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns
)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.9,
    random_state=42
)
```

```python
display(X_train.shape)
display(X_test.shape)
```

## **LINEAR REGRESSION MODEL** ##

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

```python
lr_model = LinearRegression()
```

```python
lr_model.fit(X_train, y_train)
```

```python
y_pred = lr_model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
mse = mean_squared_error(y_test, y_pred)

print("MSE:", mse)
```

```python
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

```python
r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)
```

```python
lr_comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred
})

display(lr_comparison.head(10))
```

## **RANDOM FOREST MODEL** ##

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=5,
    random_state=42
)
```

```python
rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_mse = mean_squared_error(y_test, rf_pred)
rf_rmse = np.sqrt(rf_mse)
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest MAE:", rf_mae)
print("Random Forest RMSE:", rf_rmse)
print("Random Forest R² Score:", rf_r2)
```

```python
rf_comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": rf_pred
})

display(rf_comparison.head(10))
```

```python
gb_comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": gb_pred
})

display(gb_comparison.head(10))
```

## **TEST PREDICTION** ##

```python
def predict_house_price(
    property_type,
    district,
    month_year_of_transaction_date,
    tenure,
    main_floor_area
):
    
    input_data = pd.DataFrame({
        "property_type": [property_type],
        "district": [district],
        "month_year_of_transaction_date": [month_year_of_transaction_date],
        "tenure": [tenure],
        "main_floor_area": [main_floor_area]
    })

    # One-hot encoding
    input_data = pd.get_dummies(input_data)

    # Match training columns
    input_data = input_data.reindex(
        columns=X.columns,
        fill_value=0
    )

    prediction = xgb_model.predict(input_data)

    return prediction[0]
```

```python
price = predict_house_price(
    property_type="Town House",
    district="Gombak",
    tenure="Leasehold",
    month_year_of_transaction_date="April 2022",
    main_floor_area=97
)

print("Predicted House Price: RM", round(price))
```
